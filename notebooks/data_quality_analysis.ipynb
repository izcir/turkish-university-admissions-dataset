## 4. Hesaplama ve Raporlama Anomalileri
Bu bölümde, YÖK Atlas tarafından sunulan ve bizim de yeniden hesaplayabileceğimiz bir metrikteki tutarsızlıkları inceliyoruz.

### 4.1. "Bir Kontenjana Talip Olan Aday Sayısı" Verisindeki Anomaliler

Bu bölümde, YÖK Atlas tarafından sunulan ve `total_preferences / total_quota` formülüyle hesaplanması gereken `demand_per_quota` metriğindeki tutarsızlıkları inceliyoruz. Analizimiz iki ana soruna odaklanıyor: **yuvarlama farklılıkları** ve **büyük hesaplama hataları**.

#### Adım 1: Yuvarlama Yöntemlerindeki Tutarsızlıkların Tespiti

Öncelikle, YÖK Atlas'ın kullandığı yuvarlama yönteminin yıllar içinde standart olup olmadığını anlamaya çalışıyoruz. Bunun için iki farklı yuvarlama yöntemiyle kendi oranlarımızı hesaplayıp YÖK Atlas verisiyle karşılaştırıyoruz.

```python
import numpy as np
from decimal import Decimal, ROUND_HALF_UP
import math

depts_quota_preferences = stats.merge(
    dept_preferences[['program_code', 'year', 'total_preferences', "demand_per_quota"]],
    on=['program_code', 'year'],
    how='left'
)

# Yöntem 1: Standart Python yuvarlaması (Banker's Rounding, 0.5'i en yakın çift sayıya)
depts_quota_preferences['quota_preference_ratio_round'] = round(depts_quota_preferences['total_preferences'] / depts_quota_preferences['total_quota'], 1)

# Yöntem 2: "0.5'i yukarı yuvarla" yöntemi
depts_quota_preferences['quota_preference_ratio_decimal'] = (
    depts_quota_preferences['total_preferences'] / depts_quota_preferences['total_quota']
).apply(lambda x: float(Decimal(str(x)).quantize(Decimal('0.1'), rounding=ROUND_HALF_UP)) if not (math.isnan(x) or math.isinf(x)) else None)

# Her iki yöntemle de uyuşmayanları bulalım
quota_demand_mismatch_round = depts_quota_preferences[
    (depts_quota_preferences['quota_preference_ratio_round'] != depts_quota_preferences['demand_per_quota'])
]
quota_demand_mismatch_decimal = depts_quota_preferences[
    (depts_quota_preferences['quota_preference_ratio_decimal'] != depts_quota_preferences['demand_per_quota'])
]
quota_demand_mismatch_not = depts_quota_preferences[
    (depts_quota_preferences['quota_preference_ratio_decimal'] != depts_quota_preferences['demand_per_quota']) &
    (depts_quota_preferences['quota_preference_ratio_round'] != depts_quota_preferences['demand_per_quota'])
]

print(f"Standart `round()` ile uyuşmayan kayıt sayısı: {len(quota_demand_mismatch_round)}")
print(quota_demand_mismatch_round["year"].value_counts().sort_index(), end="\n\n")
print(f"`Decimal` ('round half up') ile uyuşmayan kayıt sayısı: {len(quota_demand_mismatch_decimal)}")
print(quota_demand_mismatch_decimal["year"].value_counts().sort_index(), end="\n\n")
print(f"Her iki yöntemi de dahil ettiğimzde uyuşmayan kayıt sayısı: {len(quota_demand_mismatch_not)}")
print(quota_demand_mismatch_not["year"].value_counts().sort_index(), end="\n\n")
```

##### Bulgular ve Yorum (Yuvarlama)

Aslında bu olayı neden böyle saçma bir hatadan dolayı olduğunu anlamam biraz uzun sürdü. `round()` fonsiyonu IEEE 754 standardına göre "round half to even" yani "bankers rounding" yapıyor. Yani 0.5 olan sayıları en yakın çift sayıya yuvarlıyor. Mesela 2.5 -> 2, 3.5 -> 4 gibi. YÖK Atlas bunu hepsinde yapsa aslında sorun olmazdı ama bazı bölümlerde 2.5 olan sayıları 3'e yuvarlamış bazı bölümlerde ise 2'ye yuvarlamış. Özellikle yıllar arasında farklı kullanmış. Mesela sayılardan da anlaşılacağı üzere 2024 yılında `round()` gibi bir fonksiyon kullanmış.

`decimal` kütüphanesi ise "round half up" yani 0.5 olan sayıları hep yukarı yuvarlayan bir yuvarlama yapıyor.

**Sonuç:** YÖK Atlas'ın yuvarlama yöntemi yıllar içinde tutarsızlık göstermektedir. Bu nedenle, `demand_per_quota` sütununu kullanmak yerine bu değeri `total_preferences / total_quota` formülüyle kendiniz hesaplamanız daha güvenilir sonuçlar verecektir.


#### Adım 2: Büyük Hesaplama Hatalarının İncelenmesi

Şimdi, yuvarlama farkıyla açıklanamayan (her iki yöntemle de uyuşmayan) **2107** kaydı daha derinlemesine inceleyelim.

```python
quota_demand_mismatch_not = quota_demand_mismatch_not[[
    "program_code", "year", "total_quota", "total_preferences", 
    "quota_preference_ratio_decimal", "demand_per_quota"
]]

print("Kontenjan 0 olduğu için NaN olan kayıt sayısı:")
print(quota_demand_mismatch_not["quota_preference_ratio_decimal"].isnull().sum())

# Nanları da çıkaralım
quota_demand_mismatch_not_null = quota_demand_mismatch_not[quota_demand_mismatch_not['quota_preference_ratio_decimal'].notna()]

print("NaN'ları çıkarınca kalan kayıt sayısı:")
print(len(quota_demand_mismatch_not_null))
```

Şimdi diğer kayıtların neden farklı olduğuna bakalım.
Üstteki kısımlarda toplam tercih sayısının rank olarak verilen 1 2 3 4 .. 10+ kısmında 
verilenlerle aynı olduğunu doğruladığımız için burada sıkıntının kontenjan/yerleşme sayısında olduğunu düşündüm.
Zaten yerleşme sayılarının da 2021 yılında cinsiyet verisi ile uyumsuz olduğunu görmüştük.
Buradaki hata da ondan kaynaklanıyor olabilir. Hipotezimiz, YÖK Atlas'ın bu hesaplamada `total_enrolled` yerine
tutarlı olan `male` + `female` toplamını kullanmış olabileceğidir.

```python
quota_demand_mismatch_not_null = quota_demand_mismatch_not_null.merge(
    stats[['program_code', 'year', 'male', 'female']],
    on=['program_code', 'year'],
    how='left'
)

quota_demand_mismatch_not_null['total_enrolled_check'] = quota_demand_mismatch_not_null['male'] + quota_demand_mismatch_not_null['female']

quota_demand_mismatch_not_null['quota_preference_ratio_decimal_check'] = (
    quota_demand_mismatch_not_null['total_preferences'] / quota_demand_mismatch_not_null['total_enrolled_check']
).apply(lambda x: float(Decimal(str(x)).quantize(Decimal('0.1'), rounding=ROUND_HALF_UP)) if not (math.isnan(x) or math.isinf(x)) else None)

final_mismatch = quota_demand_mismatch_not_null[
    (quota_demand_mismatch_not_null['quota_preference_ratio_decimal_check'] != quota_demand_mismatch_not_null['demand_per_quota'])
]

print("Kontenjan yerine cinsiyet verilerinden elde edilen yerleşen sayısı ile hesaplanınca kalan uyumsuz kayıt sayısı:")
print(len(final_mismatch))
```

Bu çıkarımı da yaptığımız da 665 uyumsuz kayıt kaldı. Bunların da ne sebepten kaynaklandığını bilmiyorum.
Büyük ihtimalle dolmayan bölümler. Diğer dolan bölümlerde cinsiyet verisinde kız erkek sayısını topladığımızda

gerçek kontenjanı bulamıyoruz o verilerde sanırım.

##### Bulgular ve Yorum (Büyük Hatalar)

Büyük farklılıkların önemli bir kısmının **2021 yılındaki hatalı `total_enrolled` verisinden** kaynaklandığı görülmektedir. YÖK Atlas'ın bu metriği hesaplarken `total_enrolled` yerine daha tutarlı olan `male` + `female` toplamını kullandığı durumlarda, uyuşmayan kayıt sayısı 1786'dan **665**'e düşmektedir.

Bu durum, YÖK Atlas tarafından sunulan hesaplanmış alanlara şüpheyle yaklaşılması ve mümkün olduğunca ham metriklerden kendi hesaplamalarımızı yapmamız gerektiğini bir kez daha teyit etmektedir.

### 4.2. Tercih Sırası Detay Verilerindeki Uyumsuzluk
YÖK Atlas'ta verilen "Toplam Tercih Edilme Sayısı" ile bu tercihin kaçıncı sırada yapıldığını gösteren detayların (1., 2., 3. sıra vb.) toplamı normalde eşit olmalıdır.

```python
rank_sum = (
    dept_preferences_rank
    .groupby(['program_code', 'year'])['count']
    .sum()
    .reset_index()
    .rename(columns={'count': 'rank_sum'})
)
merge_rank = dept_preferences.merge(
    rank_sum,
    on=['program_code', 'year'],
    how='left'
)
preferences_mismatch = merge_rank[(merge_rank['total_preferences'] != merge_rank['rank_sum'])]
print(f"Toplam tercih sayısı ile detay toplamı uyuşmayan kayıt sayısı: {len(preferences_mismatch)}")

# Uyuşmazlığın nedeni: Detayların tamamen boş bırakılması
print("Uyuşmayan kayıtlardaki detay toplamı değerleri:", set(preferences_mismatch['rank_sum']))
```

Görüldüğü üzere YökAtlasta verilen toplam tercih sayısı ile 1 2 3 4 ... 10+ olarak verilen tercih sayılarının uyuşmadığı kısımlara baktığımızda sadece 0 olanların ki uyuşmuyor. Şimdi hangi yıllarda 0 oranı daha fazla ona bakalım.

```python
print("Tercih verisi eksik olan bölümlerin yıllara göre dağılımı:")
print(preferences_mismatch["year"].value_counts().sort_index(), end="\n\n")

merged_mismatch = depts.merge(
    preferences_mismatch[['program_code']],
    on='program_code',
    how='inner'
).drop_duplicates(subset=['program_code'])

print(f"Tercih verisi eksik olan bölümlerin lisans/önlisans dağılımı:")
print(f"(False: Önlisans, True: Lisans), Toplam: {merged_mismatch.shape[0]}")
undergrad_counts = merged_mismatch['is_undergraduate'].value_counts()
print(undergrad_counts, end="\n\n")

merge = stats.merge(
    depts[['program_code', 'is_undergraduate']],
    on='program_code',
    how='left'
)
year_2023_undergrad = merge[(merge['year'] == 2023) & (merge['is_undergraduate'] == False)]
print("2023 Yılındaki Toplam Önlisans Program Sayısı:")
print(year_2023_undergrad.shape[0], end="\n\n")
```

#### Bulgular ve Yorum
Analizimiz, "Toplam Tercih Sayısı" ile tercih sırası detaylarının uyuşmadığı tüm durumların, detayların **tamamen boş bırakıldığı** kayıtlardan kaynaklandığını göstermiştir. Bu durumun dağılımı incelendiğinde:
*   Uyumsuzlukların neredeyse tamamı **2023 yılına aittir**.
*   Bu kayıtların tamamı **önlisans (`is_undergraduate=False`)** programlarıdır.
*   Daha detaylı incelemede, yerleşen sayısı 0 olan 19 bölüm dışında **2023 yılındaki tüm önlisans bölümlerinin** tercih sırası detaylarının YÖK Atlas'ta boş bırakıldığı tespit edilmiştir. Bu, 2023 yılı önlisans verilerine özgü sistemsel bir eksikliktir.
